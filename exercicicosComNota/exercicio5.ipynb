```python
import numpy as np

dados = np.genfromtxt('../data/space.csv', delimiter=';', dtype=str, encoding='utf-8', skip_header=1)
```

## Questao 1

```python
status_missoes = np.char.strip(dados[:, -1])
porcentagem_sucesso = np.mean(status_missoes == 'Success') * 100
print(f"Porcentagem de missões que deram certo: {porcentagem_sucesso:.2f}%")
```

## Questao 2

```python
custos = np.char.strip(dados[:, 6]).astype(float)
custos_validos = custos[custos > 0]
media_gastos = np.mean(custos_validos)
print(f"Média de gastos das missões com valor disponível: {media_gastos:.2f}")
```

## Questao 3

```python
missoes_eua = np.sum(np.char.find(dados[:, 2], 'USA') >= 0)
print(f"Missões realizadas pelos Estados Unidos: {missoes_eua}")
```

## Questao 4

```python
spacex = dados[dados[:, 1] == 'SpaceX']
custos_spacex = np.char.strip(spacex[:, 6]).astype(float)
missao_mais_cara = spacex[np.argmax(custos_spacex)]
print(f"Missão mais cara da SpaceX: {missao_mais_cara[4]}")
print(f"Custo: {np.char.strip(missao_mais_cara[6])}")
```

## Questao 5

```python
empresas, quantidades = np.unique(dados[:, 1], return_counts=True)
for empresa, quantidade in zip(empresas, quantidades):
    print(f"{empresa}: {quantidade}")
```

## Questao 6

```python
status_foguete = np.char.strip(dados[:, 5])
porcentagem_retired = np.mean(status_foguete == 'StatusRetired') * 100
print(f"Porcentagem de missões com StatusRetired: {porcentagem_retired:.2f}%")
```

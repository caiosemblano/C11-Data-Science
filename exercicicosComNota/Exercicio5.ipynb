```python
dados = np.genfromtxt('../data/space.csv', delimiter=';', dtype=str, encoding='utf-8', skip_header=1)
status_missoes = np.char.strip(dados[:, -1])
print(status_missoes)
sucessos = status_missoes == 'Success'

porcentagem_sucesso = np.mean(sucessos) * 100

print(f"Porcentagem de missões que deram certo: {porcentagem_sucesso:.2f}%")
```

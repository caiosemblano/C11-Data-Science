```python
campo = np.zeros((2,2), dtype=int)

campo[np.random.randint(0, 2), np.random.randint(0, 2)] = 1
print(campo)

x,y = input("coordenada x [0,1]: "), input("coordenada y [0,1]: ")

jogadas_feitas = set()
perdeu = False

while len(jogadas_feitas) < 3:
    print(f"\n--- Jogada {len(jogadas_feitas) + 1} de 3 ---")
    linha = int(input("Informe a linha (0 ou 1): "))
    coluna = int(input("Informe a coluna (0 ou 1): "))

    if (linha, coluna) in jogadas_feitas:
        print("Você já escolheu essa posição. Tente outra!")
        continue

    jogadas_feitas.add((linha, coluna))

    if campo[linha, coluna] == 1:
        perdeu = True
        break
    else:
        print("Posição segura!")

# Resultado final
if perdeu:
    print("\nGame Over! :( Try Again!")
else:
    print("\nCongratulations! You beat the game! :)")
```

```python
matriz = np.random.randint(1, 10, size=(3, 5))

linhas, colunas = matriz.shape

total_elementos = linhas * colunas

if total_elementos % 2 == 0:
    paridade = "par"
else:
    paridade = "ímpar"

print("Matriz gerada:")
print(matriz)
print(f"\nDimensões: {linhas} linhas e {colunas} colunas")
print(f"Total de elementos (linhas x colunas): {total_elementos}")
print(f"Esta matriz pode se tornar um vetor unidimensional com um número {paridade} de elementos.")
```

```python
np.random.seed(10)
matriz = np.random.randint(1, 51, size=(4,4))

print(f"media das linhas: {matriz.mean(axis=1)}")
print(f"media das colunas: {matriz.mean(axis=0)}\n")

print(f"media das linhas: {matriz.mean(axis=1).max()}")
print(f"media das colunas: {matriz.mean(axis=0).max()}\n")

valores, quant = np.unique(matriz, return_counts=True)
for val, count in zip(valores, quant):
    print(f"Número {val}: {count}x")

numeros_duas_vezes = valores[quant == 2]
print(f"\nNúmeros que aparecem exatamente 2 vezes: {numeros_duas_vezes}")
```

```python
mat1 = np.zeros((3,3), int)
mat2 = np.ones((3,3), int)
mat3 = np.empty((3,3), int)
for i in range(3):
    for j in range(3):
        mat3[i][j] = (mat1[i][j] + mat2[i][j])*5
print(mat3)

print(mat3.reshape(9))
```

```python
import numpy as np

array = np.arange(10, 31, 2)

maior_valor = array.max()
indice_maior = array.argmax()

menor_valor = array.min()
indice_menor = array.argmin()

soma_elementos = array.sum()

print(f"Array gerado:\n{array}\n")
print(f"Maior valor: {maior_valor} | Índice: {indice_maior}")
print(f"Menor valor: {menor_valor} | Índice: {indice_menor}")
print(f"Soma de todos os elementos: {soma_elementos}")
```

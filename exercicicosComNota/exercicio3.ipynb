```python
n = int(input("Quantas pessoas deseja cadastrar? "))

soma_idades = 0
mulheres_menos_20 = 0

for i in range(1, n + 1):
    print(f"\n--- {i}ª Pessoa ---")
    nome = input("Nome: ")
    idade = int(input("Idade: "))
    sexo = input("Sexo [M/F]: ").strip().upper()

    soma_idades += idade

    if sexo == "F" and idade < 20:
        mulheres_menos_20 += 1

if n > 0:
    media_idade = soma_idades / n
    print("\n" + "=" * 30)
    print(f"Média de idade do grupo: {media_idade:.1f} anos")
    print(f"Mulheres com menos de 20 anos: {mulheres_menos_20}")
else:
    print("Nenhuma pessoa foi cadastrada.")
```

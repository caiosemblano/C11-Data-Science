```python
num_raiz = int(input("De qual numero você quer a taboada?"))
print("Em qual intervalo?")
inicio, fim = int(input('Inicio')), int(input("Fim"))

for i in range(inicio, fim):
    print(f'{num_raiz} * {i} = {num_raiz * i}')
```

```python
from numpy import sqrt, ceil, floor, trunc

num_decimal = float(input("Entre com um número com decimal:"))

print(sqrt(num_decimal))
print(ceil(num_decimal))
print(floor(num_decimal))
print(trunc(num_decimal))
```

```python
import re
palavra = input("qualquer palavra:")
existe_a = False
quant_vogais = 0
for char in palavra:
    if char.lower() == "a":
        existe_a = True
    print(char.upper())

    if bool(re.match(r"^[aeiouAEIOU]$", char)):
        quant_vogais += 1

print(existe_a)
print(quant_vogais)
```

```python
x, y = float(input("Primeiro num.:")), float(input("Segundo num.:"))

if (x+y) % 2 == 0:
    print(f"{x} + {y} = {x+y}, par")
else:
    print(f"{x} + {y} = {x+y}, impar")
print(x-y)
print(x*y)
print(x/y)
print(x//y)
print(x**y)
```

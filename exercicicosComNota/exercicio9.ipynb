```python
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset('iris')
iris.head()
```

```python
setosa = iris[iris['species'] == 'setosa']
setosa.shape

corr = setosa.drop(columns='species').corr()

plt.figure(figsize=(7,5))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlação entre var numéricas - Setosa')
plt.show()
```

```python
titanic = sns.load_dataset('titanic')
#titanic[['age','sex']].head()
titanic.shape
```

```python
titanic['age'].isna().sum()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=titanic, x='age', hue='sex', kde=True, bins=30, alpha=0.5)
plt.title('Distribuição da idade dos passageiros por sexo')
plt.xlabel('Idade')
plt.ylabel('Quantidade de pessoas')
```

```python
# estatĩstica extra, só testando
titanic.groupby('sex')['age'].describe()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=titanic, x='pclass', y='age', hue='sex')
plt.title('Distribuição das idades por classe e sexo - Titanic')
plt.xlabel('Classe')
plt.ylabel('Idade')
plt.legend(title='Sexo')
plt.show()
```

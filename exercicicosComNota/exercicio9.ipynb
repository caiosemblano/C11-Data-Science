# Questão 1

```python
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset('iris')
iris.head()
```

```python
setosa = iris[iris['species'] == 'setosa']
setosa.shape

corr = setosa.drop(columns='species').corr()

plt.figure(figsize=(7,5))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlação entre var numéricas - Setosa')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=titanic, x='age', hue='sex', kde=True, bins=30, alpha=0.5)
plt.title('Distribuição da idade dos passageiros por sexo')
plt.xlabel('Idade')
plt.ylabel('Quantidade de pessoas')
```

# Questão 4

```python
import seaborn as sns
import matplotlib.pyplot as plt
mpg = sns.load_dataset('mpg')
mpg[['horsepower', 'mpg']].isna().sum()
```

```python
plt.figure(figsize=(10, 6))
sns.regplot(data=mpg, x='horsepower', y='mpg',
          scatter_kws={'alpha': 0.5}, line_kws={'color': 'red'})
plt.title('Relação entre potência e consumo de combustível')
plt.xlabel('Potência (horsepower)')
plt.ylabel('Consumo (milhas por galão)')
plt.show()
```

```python
mpg['horsepower'].corr(mpg['mpg'])
```

```python
plt.figure(figsize=(10, 6))
sns.regplot(data=mpg, x='horsepower', y='mpg', order=2,
          scatter_kws={'alpha': 0.5}, line_kws={'color': 'red'})
plt.title('Relação entre potência e consumo (ajuste de 2º grau)')
plt.xlabel('Potência (horsepower)')
plt.ylabel('Consumo (milhas por galão)')
plt.show()
```

# Questão 5

```python
import pandas as pd
import numpy as np

paises = pd.read_csv('../data/paises.csv', sep=';', encoding='utf-8')

colunas = ['GDP ($ per capita)', 'Literacy (%)',
           'Infant mortality (per 1000 births)', 'Phones (per 1000)']
dados = paises[colunas]

(dados == 0).sum()
```

```python
dados = dados.replace(0, np.nan)
dados.columns = ['PIB per capita', 'Alfabetização', 'Mortalidade infantil', 'Telefones']

corr = dados.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlação entre indicadores socioeconômicos dos países')
plt.show()
```

```python
import pandas as pd

dados = pd.read_csv('../data/paises.csv', sep=';', encoding='utf-8')
print(dados)
```

# Questão 1

```python
oceania = dados[dados['Region'].str.contains('OCEANIA')]
print('Países da OCEANIA:')
print(oceania['Country'].to_string(index=False))
print(f'\nQuantidade de países da OCEANIA: {len(oceania)}')
```

# Questão 2

```python
maior_pop = dados.loc[dados['Population'].idxmax(), ['Country', 'Region', 'Population']]
print(f"País: {maior_pop['Country'].strip()}")
print(f"Região: {maior_pop['Region'].strip()}")
print(f"População: {maior_pop['Population']}")
```

# Questão 3

```python
perReg = dados.groupby('Region')
print(perReg['Literacy (%)'].mean())
```

# Questão 4

```python
sem_costa = dados[dados['Coastline (coast/area ratio)'] == 0]['Country']
print(sem_costa.to_string(index=False))
sem_costa.to_csv('noCoast.csv', index=False)
```

# Questão 5

```python
def humanitarian_help(deathrate):
    if deathrate < 9:
        return 'Balanced'
    return 'Urgent'

dados['Humanitarian Help'] = dados['Deathrate'].apply(humanitarian_help)
print(dados[['Country', 'Deathrate', 'Humanitarian Help']])
```

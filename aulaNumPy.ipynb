```python
arr = np.ones([11,11], str)
print(arr)
```

```python
for i in range(11):
    for j in range(11):
        if i == j:
            arr[i,j] = "X"
        elif 10-i == j:
            arr[i,j] = "X"
        else:
            arr[i,j] = "."

print(arr)
```
